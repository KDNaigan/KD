```python
import pandas as pd

# Read each CSV file
facebook = pd.read_csv("fb.csv")
apple = pd.read_csv("aapl.csv")
amazon = pd.read_csv("amzn.csv")
netflix = pd.read_csv("nflx.csv")
google = pd.read_csv("goog.csv")

# Add ticker
facebook["ticker"] = "FB"
apple["ticker"] = "AAPL"
amazon["ticker"] = "AMZN"
netflix["ticker"] = "NFLX"
google["ticker"] = "GOOGL"

# combine them into one dataframe
faang = pd.concat([facebook, apple, amazon, netflix, google],
                  axis=0,
                  ignore_index=True)

# save to CSV
faang.to_csv("faang.csv", index=False)

# Display
faang.head(10)
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# URL
url = "https://phapi.online/members/"

# Add headers to avoid blocking
headers = {
    "User-Agent": "Mozilla/5.0"
}

# Request page
response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text, "html.parser")

# Lists to store data
names = []
addresses = []
contacts = []

# Inspect the site structure — adjust selectors if needed
members = soup.find_all("div", class_="elementor-widget-container")

for member in members:
    text = member.get_text(separator="|", strip=True).split("|")

    # Basic parsing (adjust based on actual page structure)
    if len(text) >= 2:
        names.append(text[0])
        addresses.append(text[1])
        contacts.append(text[2] if len(text) > 2 else "Unknown")

# Create DataFrame
hospitals = pd.DataFrame({
    "name": names,
    "address": addresses,
    "contact": contacts
})

# Save CSV
hospitals.to_csv("hospitals.csv", index=False)

print(hospitals.head())
```

```python
# Load CSV
df = pd.read_csv("hospitals.csv")

# Remove duplicates
df = df.drop_duplicates()

# Clean whitespace
df["name"] = df["name"].str.strip()
df["address"] = df["address"].str.strip()
df["contact"] = df["contact"].astype(str).str.strip()

# Replace missing values
df.fillna("Unknown", inplace=True)

print(df.info())
print(df.head())
```
