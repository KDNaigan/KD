# PART 1

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
w1 = 3
b = 1
x1 = 0
```

```python
y = w1 * x1 + b
```

```python
print (y)
```

```python
x_value = np.linspace(-5, 5, 100)
y_value = w1 * x_value + b
```

```python
plt.figure()
plt.plot(x_value, y_value, label="y = 3x + 1")
plt.xlabel("X AXIS")
plt.ylabel("Y AXIS")
plt.title("LINEAR REGRESSION")
plt.show
```
