1. Create a code that demonstrate how to use recursion method to solve problem

```python
def fibonacci_r(n):
    if n<=1:
        return n
    return fibonacci_r(n-1) + fibonacci_r(n-2)
print(fibonacci_r(10))
```

2. Create a program codes that demonstrate how to use dynamic programming to solve the same problem 

```python
def fibonacci_dp(n):
    dp = [0]*(n+1)
    dp[1] = 1

    for i in range (2,n+1):
        dp[i] = dp[i-1]+dp[i-2]

        return dp[n]
print(fibonacci_dp(10))
```

##### Question: 
Explain the difference of using the recursion from dynamic programming using the given sample codes to solve the same problem

The difference between recursion and Dynamic Programming is, Recursion solves a problem by calling the same function repeatedly, it breaks the problem into smaller problems, but it does not remember the previous result. This causes the same calculations to be repeated many times, which makes it slow for larger inputs.

While Dynamic Programming is an optimization of recursion, it solves each small problems only once and store it's result. This avoids repeated calculation and make the program much faster.

3. Create a sample program codes to simulate bottom-up dynamic programming

```python
def fibonacci_BU(n):
    dp = [0] * (n+1)
    dp[0] = 0
    dp[1] = 1
    
    for i in range (2,n+1):
        dp[i] = dp[i-1] + dp[i-2]

    return dp[n]

print(fibonacci_BU(10))
```

4. Create a sample program codes to simulate tops-down dynamic programming

```python
def fibonacci_TD(n , memo={}):
    if n in memo:
        return memo[n]

    if n<=1:
        return n
    
    memo[n] = fibonacci_TD(n-1, memo) + fibonacci_TD(n-2, memo)
    return memo[n]

print(fibonacci_TD(10))

```

0/1 Knapsack Problem

* Analyze three different techniques to solve knapsacks problem
1. Recursion
2. Dynamic Programming
3. Memoization

```python
#sample code for knapsack problem using recursion
def rec_knapSack(w, wt, val, n):

  #base case
  #defined as nth item is empty;
  #or the capacity w is 0
  if n == 0 or w == 0:
    return 0

  #if weight of the nth item is more than
  #the capacity W, then this item cannot be included
  #as part of the optimal solution
  if(wt[n-1] > w):
    return rec_knapSack(w, wt, val, n-1)

  #return the maximum of the two cases:
  # (1) include the nth item
  # (2) don't include the nth item
  else:
    return max(
        val[n-1] + rec_knapSack(
            w-wt[n-1], wt, val, n-1),
            rec_knapSack(w, wt, val, n-1)
    )
```

```python
#To test:
val = [60, 100, 120] #values for the items
wt = [10, 20, 30] #weight of the items
w = 50 #knapsack weight capacity
n = len(val) #number of items

rec_knapSack(w, wt, val, n)
```

```python
#Dynamic Programming for the Knapsack Problem
def DP_knapSack(w, wt, val, n):
  #create the table
  table = [[0 for x in range(w+1)] for x in range (n+1)]

  #populate the table in a bottom-up approach
  for i in range(n+1):
    for w in range(w+1):
      if i == 0 or w == 0:
        table[i][w] = 0
      elif wt[i-1] <= w:
        table[i][w] = max(val[i-1] + table[i-1][w-wt[i-1]],
                          table[i-1][w])
  return table[n][w]
```

```python
#To test:
val = [60, 100, 120]
wt = [10, 20, 30]
w = 50
n = len(val)

DP_knapSack(w, wt, val, n)
```

```python
#Sample for top-down DP approach (memoization)
#initialize the list of items
val = [60, 100, 120]
wt = [10, 20, 30]
w = 50
n = len(val)

#initialize the container for the values that have to be stored
#values are initialized to -1
calc =[[-1 for i in range(w+1)] for j in range(n+1)]


def mem_knapSack(wt, val, w, n):
  #base conditions
  if n == 0 or w == 0:
    return 0
  if calc[n][w] != -1:
    return calc[n][w]
  
  #compute for the other cases
  if wt[n-1] <= w:
    calc[n][w] = max(val[n-1] + mem_knapSack(wt, val, w-wt[n-1], n-1),
                     mem_knapSack(wt, val, w, n-1))
    return calc[n][w]
  elif wt[n-1] > w:
    calc[n][w] = mem_knapSack(wt, val, w, n-1)
    return calc[n][w]

mem_knapSack(wt, val, w, n)
```

**Code Analysis**

Recursion: it explores all the possibility but it waste a lot of time recomputing the same subproblems.
Dynamic Programming(Bottom-Up): it solve the problem efficiently by finding the structure solution from smaller subproblems.
Memoization(Top-Down): it is more optimize recursion by storing and reusig results

Most of them can do the work but Dynamic Programming or Memoization is preffered oer a plain recursion for solving a 0/1 Knapsack Problem efficiently.

Task 2: Create a sample program that find the nth number of Fibonacci Series using Dynamic Programming

```python
def fib_series(n):
    if n <= 1:
        return n
    dp = [0] * (n + 1)
    dp[0] = 0
    dp[1] = 1
    
    for i in range(2, n+1):
        dp[i] = dp[i-1] + dp[i-2]
    
    return dp[n]

print(fib_series(10))
print(fib_series(6))
```

```python
#type your code here for recursion programming solution
#Saving Money Over Time
#Student should save some money per week
#The goal is to compute the total amount of money saved after 'n' weeks

def total_savings_recursive(n):
    # base case
    if n == 0:
        return 0
    return n + total_savings_recursive(n - 1)


weeks = 5
total_savings_recursive(weeks)
```

```python
#type your code here for dynamic programming solution
def total_savings_dp(n):
    dp = [0] * (n + 1)

    for i in range(1, n + 1):
        dp[i] = dp[i - 1] + i

    return dp[n]

total_savings_dp(weeks)
```

#### Conclusion

Dynamic Programming is an effective problem solving method that is applied to solve complex problems by breaking them down into smaller, overlapping sub problems. Rather than repeatedly finding solutions to the same subproblems, dynamic programming can remember solutions to problems and use them again, saving a huge amount of time in computation. Dynamic programming is a much better performance-wise approach to simple recursion, particularly when the size of the input is large. It can be applied with a bottom-up design, that is, the design builds solutions based on base cases, or a top-down design, that is, the design uses recursion and memoization. Altogether, dynamic programming is an effective and useful tool to solve real-life issues which need to be optimized and efficiently used.
